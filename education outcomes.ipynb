```python
import pandas as pd
riyadh = pd.read_excel('/content/Performance of students in Riyadh region.xlsx')
makkah = pd.read_excel('/content/Performance of students Makkah region.xlsx')
eastern = pd.read_excel('/content/Performance of students in Eastern region.xlsx')

riyadh['المنطقة']='الرياض'
makkah['المنطقة']='مكه المكرمه'
eastern['المنطقة']='الشرقية'

df = pd.concat([riyadh,makkah,eastern],ignore_index=True)
df.to_excel('Performance of students.xlsx')
print("عدد الصفوف:", df.shape[0])
print("عدد الاعمره:",df.shape[1])
print(df.head())
```

```python
import sqlite3

conn = sqlite3.connect('education.db')
df.to_sql('education', conn,
          if_exists='replace', index=False)

query = """
SELECT المنطقة,
       ROUND(AVG("متوسط الدرجة"), 2) as متوسط_المنطقة
FROM education
GROUP BY المنطقة
ORDER BY متوسط_المنطقة DESC
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query2 = """
SELECT المنطقة, "جنس المدرسة",
       ROUND(AVG("متوسط الدرجة"), 2) as المتوسط
FROM education
GROUP BY المنطقة, "جنس المدرسة"
ORDER BY المنطقة, المتوسط DESC
"""
print(pd.read_sql_query(query2, conn))

# أعلى 10 مدارس
query3 = """
SELECT "اسم المدرسة", المنطقة, "جنس المدرسة",
       ROUND(AVG("متوسط الدرجة"), 2) as المتوسط
FROM education
GROUP BY "اسم المدرسة"
ORDER BY المتوسط DESC
LIMIT 10
"""
print(pd.read_sql_query(query3, conn))
```

```python
# مقارنة بين العامين 1444 و1445
query4 = """
SELECT المنطقة, "العام الدراسي",
       ROUND(AVG("متوسط الدرجة"), 2) as المتوسط
FROM education
GROUP BY المنطقة, "العام الدراسي"
ORDER BY المنطقة, "العام الدراسي"
"""
print(pd.read_sql_query(query4, conn))
```
